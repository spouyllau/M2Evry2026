# Partie 3 : Visualisation (15 min)
## 1. Introduction a Matplotlib

```python
import matplotlib.pyplot as plt

annees = [1750, 1760, 1770, 1780, 1790, 1800]
population = [500, 550, 600, 650, 700, 550]

plt.figure(figsize=(10, 6))
plt.plot(annees, population, marker='o', color='b')
plt.title("Evolution population Paris 1750-1800")
plt.xlabel("Annee")
plt.ylabel("Population (mille)")
plt.grid(True)
plt.show()
```

```python
prix_ble = [0.4, 0.45, 0.6, 0.8, 1.2, 0.9]
prix_vin = [0.2, 0.25, 0.3, 0.35, 0.5, 0.4]

plt.figure(figsize=(10, 6))
plt.plot(annees, prix_ble, marker='o', label='Prix du ble')
plt.plot(annees, prix_vin, marker='s', label='Prix du vin')
plt.title("Evolution des prix a Paris")
plt.xlabel("Annee")
plt.ylabel("Prix (livres)")
plt.legend()
plt.grid(True)
plt.show()
```

## Mini-projet final (optionnel)Creez un script qui :
1. Lit un fichier texte historique
2. Nettoie le texte
3. Compte les occurrences de 3 mots cles
4. Creer un graphique simple

```python
import matplotlib.pyplot as plt

# Lire fichier
try:
    with open("archive.txt", "r", encoding="utf-8") as f:
        texte = f.read().lower()
except:
    texte = "La revolution a commence en 1789. Le roi a ete execute. La guerre a eclate."

# Nettoyer
texte = texte.strip()
texte = " ".join(texte.split())

# Compter
mots_cles = ['revolution', 'roi', 'guerre', 'paix', 'liberte']
mots = texte.split()
resultats = {}
for mot in mots_cles:
    resultats[mot] = mots.count(mot)

print("Occurrences:")
for mot, count in resultats.items():
    print(f"{mot}: {count}")

# Graphique
plt.figure(figsize=(10, 6))
plt.bar(resultats.keys(), resultats.values(), color='skyblue')
plt.title("Frequence des mots cles dans le texte")
plt.xlabel("Mots")
plt.ylabel("Occurrences")
plt.grid(True, axis='y', linestyle='--', alpha=0.7)
plt.show()
```
