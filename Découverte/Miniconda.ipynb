## Pourquoi utiliser Miniconda ?
### Explication simple
Miniconda permet d'installer **Python + des bibliothèques scientifiques** sans conflit avec d'autres logiciels.

Un **environnement** est comme une boite isolee ou vous installez uniquement ce dont vous avez besoin pour votre projet.

### Avantages :
- Pas de conflit entre differents projets
- Facile a supprimer et recreer
- Gestion simplifiee des dependances
- Parfait pour les debutants !

# Partie 1 : Telechargement et installation de Miniconda (10 min)

## A. Telecharger Miniconda
1. Allez sur : [https://docs.conda.io/en/latest/miniconda.html](https://docs.conda.io/en/latest/miniconda.html)
2. Choisissez :
   - Votre systeme d'exploitation (Windows, macOS, Linux)
   - Version Python 3.x (la plus recente)
   - Installateur graphique (pour Windows) ou version ligne de commande (pour macOS/Linux)

## B. Installer Miniconda
### Pour Windows :
1. Double-cliquez sur le fichier .exe telecharge
2. Suivez les instructions (laissez les options par defaut)
3. **Cochez** "Add Miniconda3 to my PATH environment variable" (important !)
4. Terminez l'installation

### Pour macOS/Linux :Ouvrez un terminal et executez :

````python
- Pour macOS
```bash 
Miniconda3-latest-MacOSX-x86_64.sh
```

- Pour Linux
```bash 
Miniconda3-latest-Linux-x86_64.sh
```
````

**Pendant l'installation :**
- Appuyez sur **Entree** pour lire la licence
- Tapez `yes` pour accepter
- Appuyez sur **Entree** pour confirmer le chemin d'installation
- Tapez `yes` pour initialiser Conda

## C. Verifier l'installation
Ouvrez un **nouveau terminal** (ou **Anaconda Prompt** sous Windows) et tapez :

````python
# Verifier la version de conda
conda --version
````

✅ Vous devriez voir un numero de version (ex: `conda 23.11.0`)

---
# Partie 2 : Creer un environnement pour l'histoire (5 min)

## A. Creer un environnement nommme "histoire"
Cette commande cree un environnement avec Python 3.10 :

````python
# Creer l'environnement
conda create --name histoire python=3.12
````

- `histoire` : le nom de votre environnement (vous pouvez le changer)
- `python=3.12` : la version de Python (stable et recommandee, en 2026)

**Pendant l'installation :**
- Tapez `y` pour confirmer

## B. Activer l'environnement

````python
# Activer l'environnement
conda activate histoire
````

✅ Vous devriez voir `(histoire)` apparaitre au debut de votre invite de commande.

## C. Verifier que vous etes dans le bon environnement

````python
# Verifier la version de Python
python --version
````

✅ Vous devriez voir `Python 3.10.x` (ou x est un numero de sous-version)

---
# Partie 3 : Installer Jupyter Lab (5 min)

Avec l'environnement active (`histoire`), tapez :

````python
# Installer Jupyter Lab
conda install jupyterlab
````

- Tapez `y` pour confirmer l'installation
- Attendez que l'installation se termine (quelques minutes)

## Verifier l'installation

````python
# Verifier la version de Jupyter Lab
jupyter lab --version
````

✅ Vous devriez voir un numero de version (ex: `4.0.12`)

---
# Partie 4 : Lancer Jupyter Lab (5 min)

## A. Demarrer Jupyter Lab

````python
# Lancer Jupyter Lab
jupyter lab
````

✅ Vous devriez voir un message comme :
```
To access the server, open this file in a browser:
    file:///home/user/.local/share/jupyter/runtime/jpserver-12345-open.html
Or copy and paste one of these URLs:
    http://localhost:8888/?token=...
```

## B. Ouvrir Jupyter Lab dans votre navigateur
1. Copiez le lien `http://localhost:8888/?token=...` dans votre navigateur
2. **OU** cliquez sur le fichier `.html` mentionne dans le message

✅ Jupyter Lab s'ouvre dans une nouvelle fenetre de navigateur !

## C. Decouvrir l'interface
- **A gauche** : l'arbre de fichiers (votre explorateur)
- **Au centre** : l'espace de travail
- **En haut** : la barre de menus

**Pour creer un nouveau notebook :**
1. Cliquez sur **File** → **New** → **Python 3 (ipykernel)**
2. Un nouvel onglet s'ouvre avec un notebook vide

# Problemes courants et solutions

## Probleme 1 : `conda` n'est pas reconnu (Windows)
**Solution :**
- Verifiez que vous avez coche "Add Miniconda3 to my PATH" pendant l'installation
- Redemarrez votre ordinateur
- Essayez d'utiliser **Anaconda Prompt** au lieu de l'invite de commande classique

## Probleme 2 : Port 8888 deja utilise
**Solution :**

````python
# Lancer Jupyter Lab sur un autre port
jupyter lab --port 8889
````

(Remplacez `8889` par un autre numero si necessaire)

## Probleme 3 : Jupyter Lab ne s'ouvre pas dans le navigateur
**Solution :**
- Copiez manuellement le lien `http://localhost:8888/?token=...` dans votre navigateur
- Verifiez que votre pare-feu n'a pas bloque Jupyter

## Probleme 4 : Erreur de module manquant
**Solution :**

````python
# Installer le module manquant
conda install nom_du_module
````

(Exemple : `conda install pandas`)
